# **Project Summary -**

In today’s competitive business environment, customer satisfaction is one of the most important indicators of an organization’s success. Every company strives to create a positive customer experience because satisfied customers are more likely to stay loyal, recommend the brand to others, and continue buying products or services. On the other hand, even a small percentage of dissatisfied customers can have a large impact on business growth, as they are more likely to switch to competitors, leave negative reviews, or discourage potential customers through word of mouth. For companies that handle large volumes of customer interactions every day, maintaining consistently high levels of satisfaction is a major challenge.

The dataset provided for this project represents real-world customer support interactions. It contains more than 85,000 rows of data, covering important aspects such as the channel of communication (inbound calls, outcalls, emails), the product category, the type of issue reported, timestamps of when issues were raised and responded to, handling times, agent information, and most importantly, the Customer Satisfaction (CSAT) score. This rich dataset provides a clear picture of how customers interact with support teams, what issues they face, and how those issues are handled. However, raw data by itself is not useful unless it is carefully analyzed and converted into insights that the business can act upon.

The key challenge here is that not all interactions result in good CSAT scores. While many customers report high satisfaction (score 5), a considerable number of interactions lead to lower scores (1, 2, or 3). These low scores indicate dissatisfaction, which can be caused by a variety of factors such as slow response times, poor product quality, long handling times, or differences in performance across channels and agents. Identifying the exact reasons for dissatisfaction is difficult because customer support data is large, complex, and contains both structured (numbers, categories) and unstructured data (customer remarks). Without systematic analysis, the company cannot pinpoint the drivers of dissatisfaction or take focused actions to improve them.

Therefore, the main problem this project aims to address is: how can we analyze customer support interactions to find the factors that drive dissatisfaction, and how can we use machine learning to predict whether a customer is likely to be satisfied or dissatisfied based on these factors? Answering this question will not only help the company understand past performance but also provide tools to predict future outcomes and act proactively.

Another important challenge is data imbalance. The dataset is heavily skewed towards high CSAT scores, meaning that most customers are satisfied, while only a smaller group reports dissatisfaction. While this is good news for the business overall, it creates difficulty for machine learning models. If not handled properly, the model may simply predict “satisfied” most of the time and still achieve high accuracy, but it would fail to identify the truly dissatisfied customers, which are the most important cases from a business perspective. This project must therefore deal with imbalanced data carefully, using techniques like resampling and appropriate evaluation metrics.

From a business perspective, solving this problem is very impactful. If the company can reduce response times, optimize handling times, and train agents based on insights from data, it can improve customer satisfaction even further. High-value product categories like electronics and mobiles, which currently show lower CSAT, can receive special attention. Channels like email, which often perform worse than phone-based interactions, can be improved through better resource allocation or automation. By taking such targeted actions, the company can prevent customer churn, reduce returns, and increase brand loyalty.

In conclusion, the problem addressed by this project is both analytical and business-focused. Analytically, it involves cleaning, preparing, and modeling a complex dataset to extract meaningful insights. From a business point of view, it involves understanding how customer satisfaction is influenced by different factors and using predictive models to take early action on at-risk cases. Successfully addressing this problem will help the company not only maintain but also improve its reputation for customer service, directly leading to long-term business growth and customer loyalty.

## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# 1. Drop duplicates
df.drop_duplicates(inplace=True)
````

````python
[ ]
# 2. Standardize column names (remove spaces, lowercase, replace non-alphanumerics with underscore)
df.columns = df.columns.str.strip().str.lower().str.replace(r'[^0-9a-z]+', '_', regex=True)
````

````python
[ ]
# 3. Handle missing values (categorical -> 'Unknown', numeric -> median) using .loc for safety
cat_cols = df.select_dtypes(include=['object', 'category']).columns.tolist()
num_cols = df.select_dtypes(include=[np.number]).columns.tolist()

for col in cat_cols:
    if df.loc[:, col].isnull().any():
        df.loc[:, col] = df.loc[:, col].fillna('Unknown')

for col in num_cols:
    if df.loc[:, col].isnull().any():
        df.loc[:, col] = df.loc[:, col].fillna(df.loc[:, col].median())
````

````python
[ ]
# 4. Convert date columns to datetime format (try common explicit format first, fallback to flexible parse)
date_cols = ['order_date_time', 'issue_reported_at', 'issue_responded', 'survey_response_date']

for col in date_cols:
    if col in df.columns:
        # try the exact format many rows appear to use (e.g., "01/08/2023 11:13")
        df.loc[:, col] = pd.to_datetime(df.loc[:, col], format="%d/%m/%Y %H:%M", errors='coerce')
        # if majority failed, fallback to flexible parsing (avoids heavy pandas infer warning on entire column)
        if df[col].isna().mean() > 0.5:
            df.loc[:, col] = pd.to_datetime(df.loc[:, col], errors='coerce')
````

````python
[ ]
# 5. Create new feature: response_time_hours (issue_responded - issue_reported_at) in hours
# Convert to datetime
df['issue_responded'] = pd.to_datetime(df['issue_responded'], errors='coerce')
df['issue_reported_at'] = pd.to_datetime(df['issue_reported_at'], errors='coerce')

# Calculate response time in hours
if ('issue_responded' in df.columns) and ('issue_reported_at' in df.columns):
    df['response_time_hours'] = (
        (df['issue_responded'] - df['issue_reported_at']).dt.total_seconds() / 3600.0
    )
else:
    df['response_time_hours'] = np.nan

````

````python
[ ]
# 6. Remove irrelevant columns safely (like unique_id)
df.drop(columns=['unique_id'], inplace=True, errors='ignore')
````

## ***5. Hypothesis Testing***

### Based on your chart experiments, define three hypothetical statements from the dataset. In the next three questions, perform hypothesis testing to obtain final conclusion about the statements through your code and statistical testing.

Three hypotheses I tested were:

Response time has an effect on CSAT scores.

CSAT differs across product categories.

CSAT differs between customer support channels.

### Hypothetical Statement - 1

#### 1. State Your research hypothesis as a null hypothesis and alternate hypothesis.

Null hypothesis (H0): Response time does not affect CSAT.

Alternate hypothesis (H1): Response time affects CSAT.

#### 2. Perform an appropriate statistical test.

````python
# Perform Statistical Test to obtain P-Value


import scipy.stats as stats

# drop missing values for relevant columns
anova_data = df[['channel_name', 'csat_score']].dropna()

# group CSAT scores by channel
groups = [group['csat_score'].values for name, group in anova_data.groupby('channel_name')]

# perform one-way ANOVA
f_stat, p_value = stats.f_oneway(*groups)

print("ANOVA F-statistic:", f_stat)
print("ANOVA p-value:", p_value)
````

##### Which statistical test have you done to obtain P-Value?

I used an ANOVA test because I wanted to check if there were significant differences in CSAT across different groups of response time.

````python
# Handling Outliers & Outlier treatments

[ ]
# 1. Choose numeric columns to treat (exclude target 'csat_score' if needed)
num_cols = df.select_dtypes(include=[np.number]).columns.tolist()
if 'csat_score' in num_cols:
    num_cols.remove('csat_score')
print("Numeric columns to consider:", num_cols)

# 2. IQR capping function and application (create new *_capped columns)
def iqr_cap_series(s: pd.Series, k: float = 1.5) -> pd.Series:
    q1 = s.quantile(0.25)
    q3 = s.quantile(0.75)
    iqr = q3 - q1
    lower = q1 - k * iqr
    upper = q3 + k * iqr
    return s.clip(lower=lower, upper=upper)

# columns you want to cap - adjust list as needed (example)
cols_to_treat = [c for c in ['item_price', 'response_time_hours', 'connected_handling_time'] if c in df.columns]

for c in cols_to_treat:
    df.loc[:, c + '_capped'] = iqr_cap_series(df.loc[:, c].astype(float))
    print(f"Created: {c + '_capped'} (non-null: {df[c].notna().sum()})")

# 3. Log transform for strictly non-negative columns (create *_log1p)
cols_to_log = []
for c in cols_to_treat:
    # require at least some positive values
    vals = df.loc[:, c].dropna()
    if not vals.empty and (vals > 0).any():
        cols_to_log.append(c)

for c in cols_to_log:
    df.loc[:, c + '_log1p'] = np.log1p(df.loc[:, c].astype(float).clip(lower=0))
    print("Created log1p:", c + '_log1p')

# 4. Z-score flagging (mark extremes, do NOT drop automatically) - using original values
z_thresh = 3.0
for c in cols_to_treat:
    col_z = c + '_zscore'
    out_flag = c + '_is_outlier'
    # compute zscore only where not-null
    if df[c].notna().sum() > 1:
        df.loc[:, col_z] = stats.zscore(df.loc[:, c].astype(float).fillna(df.loc[:, c].median()))
        df.loc[:, out_flag] = df.loc[:, col_z].abs() > z_thresh
        print(f"{c}: flagged {df[out_flag].sum()} outliers with |z| > {z_thresh}")
    else:
        df.loc[:, col_z] = np.nan
        df.loc[:, out_flag] = False


# 6. Robust scaling (for modeling) - create *_robust_scaled columns
robust_cols = [c for c in cols_to_treat if c in df.columns]
if robust_cols:
    scaler = RobustScaler()
    scaled = scaler.fit_transform(df.loc[:, robust_cols].fillna(df.loc[:, robust_cols].median()))
    for i, c in enumerate(robust_cols):
        df.loc[:, c + '_robust_scaled'] = scaled[:, i]
    print("Added robust-scaled columns:", [c + '_robust_scaled' for c in robust_cols])
else:
    print("No columns to robust-scale.")



# 7. Quick before/after summary for treated columns
for c in cols_to_treat:
    print("\nColumn:", c)
    print(" original min, q1, median, q3, max ->",
          df[c].min(), df[c].quantile(0.25), df[c].median(), df[c].quantile(0.75), df[c].max())
    if c + '_capped' in df.columns:
        print(" capped    min, q1, median, q3, max ->",
              df[c + '_capped'].min(), df[c + '_capped'].quantile(0.25), df[c + '_capped'].median(),
              df[c + '_capped'].quantile(0.75), df[c + '_capped'].max())
    if c + '_log1p' in df.columns:
        print(" log1p mean:", df[c + '_log1p'].mean())
    if c + '_is_outlier' in df.columns:
        print(" outlier count:", df[c + '_is_outlier'].sum())
````

````python
[ ]
# Fixed fast hyperparameter tuning (HalvingRandomSearchCV) with a safe fallback to small RandomizedSearchCV
import time
from sklearn.experimental import enable_halving_search_cv  # noqa
from sklearn.model_selection import HalvingRandomSearchCV, RandomizedSearchCV

# Choose estimator (prefer HistGradientBoosting; fallback to RandomForest if not available)
try:
    from sklearn.ensemble import HistGradientBoostingClassifier
    tune_estimator = HistGradientBoostingClassifier(random_state=42)
    # NOTE: do NOT include 'max_iter' here because resource must not be a searched param
    param_dist = {
        "learning_rate": [0.01, 0.05, 0.1],
        "max_leaf_nodes": [15, 31, 63, None],
        "max_depth": [3, 5, None],
        "min_samples_leaf": [20, 50, 100]
    }
    estimator_name = "HistGradientBoosting"
except Exception:
    from sklearn.ensemble import RandomForestClassifier
    tune_estimator = RandomForestClassifier(random_state=42, class_weight='balanced', n_jobs=-1)
    param_dist = {
        "n_estimators": [100, 200, 400],
        "max_depth": [None, 6, 12],
        "min_samples_split": [2, 5, 10],
        "min_samples_leaf": [1, 2, 4],
        "max_features": ["sqrt", 0.5]
    }
    estimator_name = "RandomForest"

# Try HalvingRandomSearchCV (successive halving) first — fast search
try:
    halving = HalvingRandomSearchCV(
        estimator=tune_estimator,
        param_distributions=param_dist,
        factor=3,
        resource='n_samples',    # use sample size as resource (NOT part of param_dist)
        scoring='f1',
        cv=3,
        random_state=42,
        n_jobs=-1,
        verbose=2
    )
    t0 = time.time()
    halving.fit(X_train, y_train)
    elapsed = time.time() - t0
    print(f"HalvingRandomSearchCV finished in {elapsed:.1f}s")
    print("Best params:", halving.best_params_)
    print("Best CV score:", halving.best_score_)
    best_estimator = halving.best_estimator_

# Fallback: if Halving raises an error or is slow, run a small RandomizedSearch for a quick result
except Exception as e:
    print("HalvingRandomSearchCV failed or raised an exception:", str(e))
    print("Falling back to a SMALL RandomizedSearchCV (n_iter=12, cv=3) to return results quickly.")
    rnd = RandomizedSearchCV(
        estimator=tune_estimator,
        param_distributions=param_dist,
        n_iter=12,            # small, fast search
        scoring='f1',
        cv=3,
        random_state=42,
        n_jobs=-1,
        verbose=2
    )
    t0 = time.time()
    rnd.fit(X_train, y_train)
    elapsed = time.time() - t0
    print(f"RandomizedSearchCV (small) finished in {elapsed:.1f}s")
    print("Best params (randomized):", rnd.best_params_)
    print("Best CV score (randomized):", rnd.best_score_)
    best_estimator = rnd.best_estimator_

# After tuning: evaluate on test set
try:
    from sklearn.metrics import classification_report, roc_auc_score
    y_pred = best_estimator.predict(X_test)
    try:
        y_proba = best_estimator.predict_proba(X_test)[:,1]
        auc = roc_auc_score(y_test, y_proba)
    except Exception:
        y_proba = None
        auc = float('nan')
    print("\nTest set classification report (best_estimator):")
    print(classification_report(y_test, y_pred, zero_division=0))
    print("Test ROC AUC (if available):", auc)
except Exception as eval_err:
    print("Evaluation failed:", eval_err)
````
